```python
# Find service fees outside the IQR outlier limits.
def find_outliers(df, service_fee):
    Q1 = df["service_fee"].quantile(0.25)
    Q3 = df["service_fee"].quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    return df[(df["service_fee"] < lower) | (df["service_fee"] > upper)]

find_outliers(df, "service_fee")
```

```python
# One row per driver, then compare rating with late rate
drivers = done.groupby("driver_id").agg(rating=("delivery_partner_rating", "first"),
                                        experience=("delivery_partner_experience_months", "first"),
                                        late_rate=("late_delivery", "mean"),
                                        orders=("order_id", "count"))
sns.scatterplot(data=drivers, x="rating", y="late_rate", color="BLUE")
plt.title("Each dot is a driver: rating vs late rate")
plt.show()
print(drivers[["rating", "experience", "late_rate"]].corr().round(2))
```
